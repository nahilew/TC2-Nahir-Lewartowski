Polo en DC ($\omega = 0$): Impedancia tiende a $\infty$ (comportamiento capacitivo a muy baja frecuencia).   

Cero en $\omega_z1 = 1\text{ rad/s}$ ($f_{z1} \approx 159\text{ mHz}$): Mínimo local de $\vert{}Z\vert{}$.   

Polo en $\omega_p1 = \sqrt{2} \approx 1.414\text{ rad/s}$ ($f_{p1} \approx 225\text{ mHz}$): Máximo local de $\vert{}Z\vert{}$.   

Cero en $\omega_z2 = \sqrt{3} \approx 1.732\text{ rad/s}$ ($f_{z2} \approx 276\text{ mHz}$): Segundo mínimo local de $\vert{}Z\vert{}$.   

Polo en $\omega \to \infty$: $\vert{}Z\vert{} \to \infty$ (comportamiento puramente inductivo).



Fase:

Al ser una red LC pura sin pérdidas, la impedancia es una reactancia pura $Z(j\omega) = j X(\omega)$:

En los intervalos donde es capacitiva ($X(\omega) < 0$), la fase es $-90^\circ$.En los intervalos donde es inductiva ($X(\omega) > 0$), la fase es $+90^\circ$.

En cada cruce por cero o polo, la fase conmuta bruscamente entre $-90^\circ$ y $+90^\circ$.

Pendiente (Ley de variación): Por el teorema de reactancia de Foster, para toda función inmitancia LC se cumple: $\frac{dX(\omega)}{d\omega} > 0$

Es decir, los ceros y polos se intercalan de manera alternada sobre el eje $j\omega$.

$$Y(s) = \frac{s^5 + 18s^3 + 48s}{6s^4 + 42s^2 + 48}$$

$$Z(s) = \frac{6s^4 + 42s^2 + 48}{s(s^4 + 18s^2 + 48)}$$

$$k_0 = \left. s \cdot Z(s) \right\vert{}_{s = 0} = \left. \frac{6s^4 + 42s^2 + 48}{s^4 + 18s^2 + 48} \right\vert{}_{s = 0} = 1$$

$$\frac{k_0}{s} = \frac{1}{s} \implies {C_1 = 1\text{ F}}$$

$$Z_a(s) = Z(s) - \frac{1}{s C_1} = \frac{6s^4 + 42s^2 + 48}{s(s^4 + 18s^2 + 48)} - \frac{1}{s} = \frac{5s^2 + 24}{s^4 + 18s^2 + 48}$$

$$Y_a(s) = \frac{1}{Z_a(s)} = \frac{s^4 + 18s^2 + 48}{s(5s^2 + 24)}$$

$$Z_{\text{tanque}}(s) = \frac{A \cdot s}{s^2 + \omega_0^2}$$

$$k_0^{(Y)} = \left. s \cdot Y_a(s) \right\vert{}_{s = 0} = \left. \frac{s^4 + 18s^2 + 48}{5s^2 + 24} \right\vert{}_{s = 0} = 2$$

$$\frac{1}{s L_1} = \frac{2}{s} \implies {L_1 = \frac{1}{2}\text{ H}}$$

$$Y_b(s) = Y_a(s) - \frac{2}{s} = \frac{s^4 + 18s^2 + 48}{s(5s^2 + 24)} - \frac{2}{s} = \frac{s(s^2 + 8)}{5s^2 + 24}$$

$$Z_b(s) = \frac{5s^2 + 24}{s(s^2 + 8)}$$

$$Z_b(s) = Z_{\text{tanque}}(s) + Z_{C_3}(s) = \frac{A \cdot s}{s^2 + 8} + \frac{1}{s C_3}$$

$$k_0^{(Z)} = \left. s \cdot Z_b(s) \right\vert{}_{s = 0} = \left. \frac{5s^2 + 24}{s^2 + 8} \right\vert{}_{s = 0} = \frac{24}{8} = 3$$

$${C_3 = \frac{1}{3}\text{ F}}$$

$$Z_{\text{tanque}}(s) = Z_b(s) - \frac{3}{s} = \frac{5s^2 + 24}{s(s^2 + 8)} - \frac{3}{s} = \frac{2s}{s^2 + 8}$$

$$\frac{1}{C_2} = 2 \implies {C_2 = \frac{1}{2}\text{ F}}$$

$$\frac{1}{L_2 C_2} = 8 \implies {L_2 = \frac{1}{4}\text{ H}}$$

En DC ($\omega = 0$), los inductores actúan como cortocircuitos y los capacitores como circuitos abiertos. Mirando el circuito, el primer componente que encuentra la señal es el capacitor $C_1$ conectado en estricta serie a la entrada. Al comportarse como un circuito abierto, $C_1$ bloquea totalmente el paso de corriente continua hacia el resto de la red. Por lo tanto, la impedancia de entrada es infinita, lo que equivale a una admitancia nula.   

Si evaluamos la expresión analítica de la admitancia dada en el enunciado en $s = 0$:
   $$Y(0) = \frac{0^5 + 18(0)^3 + 48(0)}{6(0)^4 + 42(0)^2 + 48} = 0$$La función presenta un cero en el origen, lo que confirma que la admitancia es nula en DC. En el gráfico de LTspice se observa esto como una pendiente ascendente de $+20\text{ dB/década}$ para frecuencias muy bajas, arrancando desde valores fuertemente negativos en decibelios, lo que indica que a medida que se acerca a $0\text{ Hz}$, $Y(s) \to 0$.

$$Z_{eq}(s) = R_1 + \frac{s R_2}{s + \frac{R_2}{L_2}} + \frac{\frac{1}{C_1}}{s + \frac{1}{R_3 C_1}}$$

$$Z(s) = R_1 + \frac{K_1 s}{s + 5} + \frac{K_2}{s + 1} = \frac{s^2+6s+8}{(s+1)(s+5)}$$

$$Z(\infty) = R_1 + K_1 + 0 = 1 \implies {K_1 = 1 - R_1}$$

$$Z(0) = R_1 + 0 + \frac{K_2}{1} = \frac{8}{5} = 1.6 \implies {K_2 = 1.6 - R_1}$$

$$\text{Numerador} = R_1(s^2+6s+5) + K_1 s (s+1) + K_2 (s+5)$$

$$6 R_1 + K_1 + K_2 = 6$$

$$6 R_1 + (1 - R_1) + (1.6 - R_1) = 6 \implies {R_1 = 0.85\ \Omega}$$

$$K_1 = 1 - 0.85 = {0.15}$$

$$K_2 = 1.6 - 0.85 = {0.75}$$

$$\frac{s R_2}{s + \frac{R_2}{L_2}} = \frac{0.15 s}{s + 5}$$

$$R_2 = 0.15\ \Omega$$

$$L_2 = 0.03\text{ H}$$

$$\frac{\frac{1}{C_1}}{s + \frac{1}{R_3 C_1}} = \frac{0.75}{s + 1}$$
